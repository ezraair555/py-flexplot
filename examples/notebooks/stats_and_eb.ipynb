# Biostatistical Tools with py-flexplot

This notebook covers the `fifer` and `ebbr` ports within `py-flexplot`, focusing on data cleanup, model comparison, and Empirical Bayes shrinkage.

```
import pandas as pd
import numpy as np
from pyflexplot import model_comparison, fit_beta_prior, add_ebb_estimate, eliminated_columns, p_format
import statsmodels.formula.api as smf
```

## 1. Data Cleanup: eliminated_columns

Automatically drop columns with high missingness.

```
df_missing = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [np.nan, np.nan, np.nan, 5],
    'C': [1, 2, 3, 4]
})
# Drop columns with > 50% missing data
clean_df = eliminated_columns(df_missing, threshold=0.5)
clean_df
```

## 2. Model Comparison

Compare a baseline model against an augmented model.

```
n = 100
data = pd.DataFrame({
    'y': np.random.normal(size=n),
    'x1': np.random.normal(size=n),
    'x2': np.random.normal(size=n)
})

model_small = smf.ols('y ~ x1', data=data).fit()
model_large = smf.ols('y ~ x1 + x2', data=data).fit()

fit_stats, p_val = model_comparison(model_small, model_large)
print(fit_stats)
print(f'\nLRT P-value: {p_format(p_val)}')
```

## 3. Empirical Bayes Shrinkage (The ebbr Port)

Estimating stable rates from noisy binomial data.

```
# Case study: Website Conversion Rates
sites = pd.DataFrame({
    'conversions': [2, 15, 120, 1],
    'visitors': [10, 100, 1000, 50]
})

# Shrink the raw rates toward the aggregate prior
shrunken = add_ebb_estimate(sites, 'conversions', 'visitors')
shrunken[['conversions', 'visitors', 'ebb_fitted', 'ebb_low', 'ebb_high']]
```
