# Case Study: Titanic Survival Analysis with py-flexplot

This notebook demonstrates the power of `flexplot`'s intelligent formula syntax using the famous Titanic dataset. We will explore survival patterns across age, gender, and passenger class with zero manual geom configuration.

```
import pandas as pd
import numpy as np
from pyflexplot import flexplot
from plotnine.data import titanic
import matplotlib.pyplot as plt

# Load dataset (provided by plotnine)
df = titanic.copy()
# Map 'survived' to numeric for smoother probability plotting if needed
df['survived_num'] = np.where(df['survived'] == 'yes', 1, 0)
df.head()
```

## 1. Univariate: Survival by Class

We start with a simple question: Did passenger class affect survival rates? 

Formula: `survived_num ~ class`

```
p1 = flexplot('survived_num ~ class', data=df)
p1
```

**Narrative**: `flexplot` detects that `class` is categorical and `survived_num` is numeric. It automatically generates a jittered dot plot and overlays bootstrapped means and confidence intervals. We clearly see the 'social gradient' of survival, where 1st class passengers had significantly higher survival probabilities than 3rd class.

## 2. Bivariate: Class and Sex

'Women and children first' was the protocol. Let's see how gender interacted with class.

Formula: `survived_num ~ class + sex`

```
p2 = flexplot('survived_num ~ class + sex', data=df)
p2
```

**Narrative**: By adding `+ sex` to the formula, `flexplot` automatically maps gender to color and groups the means. The visualization reveals that while women (higher means) survived at higher rates across all classes, the class disparity was much more lethal for men.

## 3. Multivariate: Faceting by Age

Finally, let's look at the impact of age. Was the class/gender advantage consistent for both children and adults?

Formula: `survived_num ~ sex + class | age`

```
p3 = flexplot('survived_num ~ sex + class | age', data=df)
p3
```

**Narrative**: Using the `|` syntax, we facet the plot by age. `flexplot` creates two panels (Child and Adult). We can now see that for children, survival was very high in 1st and 2nd class, but dropped significantly in 3rd class. For adults, the gender gap (women vs men) is the dominant feature across all classes.

## 4. Interaction Summary

The previous plots show a clear three-way interaction: survival was a function of your class, your gender, and your age. `py-flexplot` allowed us to uncover these complex relationships with just a few lines of formula-based code, handled the statistical summaries (bootstrapping) automatically, and maintained a clean, publication-ready look via `plotnine`.
