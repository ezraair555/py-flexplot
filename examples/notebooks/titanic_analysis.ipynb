# Case Study: Titanic Survival Analysis with py-flexplot

This notebook demonstrates the power of `flexplot`'s intelligent formula syntax using a simulated Titanic dataset. We will explore survival patterns across age, gender, and passenger class with zero manual geom configuration.

```
import pandas as pd
import numpy as np
from pyflexplot import flexplot
import matplotlib.pyplot as plt

# Create simulated Titanic data
np.random.seed(42)
n = 500
data = {
    'class': np.random.choice(['1st', '2nd', '3rd'], size=n, p=[0.25, 0.25, 0.50]),
    'sex': np.random.choice(['male', 'female'], size=n),
    'age': np.random.choice(['Adult', 'Child'], size=n, p=[0.85, 0.15])
}
df = pd.DataFrame(data)

# Define survival probabilities based on historical patterns
def get_prob(row):
    p = 0.3 # base
    if row['sex'] == 'female': p += 0.4
    if row['class'] == '1st': p += 0.2
    if row['class'] == '2nd': p += 0.1
    if row['age'] == 'Child' and row['class'] != '3rd': p += 0.2
    return np.clip(p, 0, 1)

df['survived_num'] = df.apply(lambda row: np.random.binomial(1, get_prob(row)), axis=1)
df.head()
```

## 1. Univariate: Survival by Class

Formula: `survived_num ~ class`

```
p1 = flexplot('survived_num ~ class', data=df)
p1
```

**Narrative**: `flexplot` detects that `class` is categorical and `survived_num` is numeric. It automatically generates a jittered dot plot and overlays bootstrapped means and confidence intervals.

## 2. Bivariate: Class and Sex

Formula: `survived_num ~ class + sex`

```
p2 = flexplot('survived_num ~ class + sex', data=df)
p2
```

**Narrative**: Mapping `sex` to color reveals the gender interaction across classes.

## 3. Multivariate: Faceting by Age

Formula: `survived_num ~ sex + class | age`

```
p3 = flexplot('survived_num ~ sex + class | age', data=df)
p3
```

**Narrative**: Faceting by `age` uncovers survival priority for children in higher classes.

## 4. Interaction Summary

`py-flexplot` handles the statistical summaries (bootstrapping) automatically while maintaining a clean, publication-ready look via `plotnine`.
