# py-flexplot Quickstart: Intelligent Data Visualization

This notebook demonstrates the core functionality of `py-flexplot`, a Python port of Dustin Fife's visualization and statistical tools. `py-flexplot` uses a formula-based syntax to automatically determine the most appropriate plot type for your data, similar to the original R implementation.

```
import pandas as pd
import numpy as np
from pyflexplot import flexplot, added_plot, model_comparison
import statsmodels.formula.api as smf
from plotnine import *

# Set seed for reproducibility
np.random.seed(42)
```

## 1. Load Sample Data

We'll use a synthetic dataset with numeric and categorical variables.

```
n = 200
df = pd.DataFrame({
    'y': np.random.normal(size=n),
    'x_num': np.random.normal(size=n),
    'x_cat': np.random.choice(['Group A', 'Group B', 'Group C'], size=n),
    'facet_var': np.random.choice(['Region 1', 'Region 2'], size=n)
})
# Add some relationship
df['y'] += 0.5 * df['x_num']
df.head()
```

## 2. Basic Intelligent Plotting

### Quantitative Predictor
`flexplot` automatically detects that `x_num` is numeric and creates a scatterplot with a regression line.

```
p = flexplot('y ~ x_num', data=df)
p
```

### Categorical Predictor
`flexplot` detects that `x_cat` is categorical and creates a jittered dot plot with mean and bootstrapped confidence intervals.

```
p = flexplot('y ~ x_cat', data=df)
p
```

## 3. Multivariate and Faceted Plots

### Color Grouping
Adding a second predictor after the `+` sign automatically maps it to color/grouping.

```
p = flexplot('y ~ x_num + x_cat', data=df)
p
```

### Faceting (The 'Given' Syntax)
Use the `|` symbol to define faceting variables.

```
p = flexplot('y ~ x_num | facet_var', data=df)
p
```

## 4. Added Variable Plots (Partial Regression)

Use `added_plot` to visualize the relationship between `y` and `x_num` after controlling for `x_cat`.

```
p = added_plot('y ~ x_num + x_cat', data=df)
p
```

## 5. Statistical Tools: Model Comparison

Compare the fit of a simple model vs. a model with multiple predictors.

```
model1 = smf.ols('y ~ x_num', data=df).fit()
model2 = smf.ols('y ~ x_num + x_cat', data=df).fit()

res, p_val = model_comparison(model1, model2)
print(res)
print(f'\nP-value for difference: {p_val:.4f}')
```
